```python
import sqlite3

connection = sqlite3.connect("database.db")
```

```python
cursor = connection.cursor()
```

```python
## Create a table

connection.execute(
    '''
    Create Table If Not Exists employee(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department Text
    )
'''
)

##Commit Changes
connection.commit()
```

```python
## Insert data in sqlite database

cursor.execute(
    '''
    INSERT INTO employees(name, age, department)
    VALUES('Tejas', 19, 'Student')
'''
)

cursor.execute(
    '''
    INSERT INTO employees(name, age, department)
    VALUES('Shardul', 112, 'Dead')
'''
)

cursor.execute(
    '''
    INSERT INTO employees(name, age, department)
    VALUES('Prashant', 55, 'Senior Dev')
'''
)


connection.commit()
```

```python
## Query the data from the table

cursor.execute('Select * from employees')
employees = cursor.fetchall()

## Print the queried data

for employee in employees:
    print(employee)
```

```python
## Update the data in the table

cursor.execute(
    '''
    Update employees
    Set age = 132
    Where name='Shardul'
    '''
)

connection.commit()

```

```python
cursor.execute('Select * from employees')
employees = cursor.fetchall()

for employee in employees:
    print(employee)
```

```python
## Delete data from table

cursor.execute(
    '''
    delete from employees
    where id = 2
'''
)

connection.commit()
```
