```python
df_filled = df.fillna(0)
```

```python
## Filling missing values with the mean of the column

df['Quantity_Sold_fillNA'] = df['Quantity_Sold'].fillna(df['Quantity_Sold'].mean())
df
```

```python
## Renaming Columns

df.rename(columns={'Discount': 'Sales Discount'})
```

```python
df = df.rename(columns={'Sales Discount': 'Discount'})
df.head(5)
```

```python
## Change datatype
df['Product_ID_New'] = df['Product_ID'].astype(float)
df.head(5)
```

```python
df['Unit Cost New'] = df['Unit_Cost'].fillna(df['Unit_Cost'].mean()).astype(int)
df.head()
```

```python
df['New Discount'] = df['Discount'].apply(lambda x:x*2)
df.head()
```

```python
## Data Aggregating And Grouping
df.head()
```

```python
grouped_mean = df.groupby('Product_Category')['Unit_Price'].mean()
print(grouped_mean)
```

```python
## Aggregate multiple functions

grouped_agg = df.groupby('Region')['Unit_Price'].agg(['mean', 'sum', 'count'])
grouped_agg
```

```python
## Merging and Joining Dataframes
df1 = pd.DataFrame({'key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
df1
```

```python
df2
```

```python
## Merge Dataframes on a key column
pd.merge(df1, df2, on='key', how='inner')
```

```python
pd.merge(df1, df2, on='key', how='outer')
```

```python
pd.merge(df1, df2, on='key', how='left')
```

```python
pd.merge(df1, df2, on='key', how='right')
```
