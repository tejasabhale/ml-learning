```python
import pandas as pd

## Series
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series \n", series)
```

```python
## Create a series from dictionary

data = {'a': 1, 'b': 2, 'c': 3}

series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10, 20, 30]
index = ['a', 'b', 'c']
pd.Series(data, index = index)
```

```python
## Dataframes
## Create a dataframe from dictionary of list

data = {
    'Name': ['Tejas', 'John', 'Jack'],
    'Age': [19, 30, 45],
    'City': ['Pune', 'New York', 'Florida']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
data = [
    {'Name': 'Tejas', 'Age': 19, 'City': 'Pune'},
    {'Name': 'Rahul', 'Age': 19, 'City': 'Pune'},
    {'Name': 'Sameer', 'Age': 19, 'City': 'Pune'},
    {'Name': 'Pramod', 'Age': 19, 'City': 'Pune'},
]

df = pd.DataFrame(data)
print(df)
print(type(df))
```
